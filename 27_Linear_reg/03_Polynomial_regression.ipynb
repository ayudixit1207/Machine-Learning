```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
x=6*np.random.rand(100,1)-3
y=0.5*x**2+1.5*x+2+np.random.randn(100,1)
## quadratic equation used y=0.5x^2+1.5x+2+outliers
plt.scatter(x,y,color='blue')
```

```python
from sklearn.model_selection import train_test_split
x_train,y_train,x_test,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
## lets implement simple linea rregression
from sklearn.linear_model import LinearRegression
regressionn=LinearRegression()
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
regressionn.fit(x_train, y_train)
```

```python
from sklearn.metrics import r2_score
score=r2_score(y_test,regressionn.predict(x_test))
print("R2 score for linear regression is:",score)

```

```python
## visuaize this model
plt.plot(x_train, regressionn.predict(x_train), color='red')
plt.scatter(x_train,y_train,color='blue')
plt.xlabel('xdataset')
plt.ylabel('ydataset')
plt.title('Polynomial Regression')
plt.show()
```

   Lets apply polynomial transformation

```python
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=2,include_bias=True)

#x=5

# [1,5,25]
```

include_bias=True → har row ke start me 1 add karo, taaki model intercept/bias learn kar sake.


```python
x_train_poly=poly.fit_transform(x_train)
x_test_poly=poly.transform(x_test)

#Convert x_train into polynomial features, then apply the same transformation to x_test.
```

```python
x_train_poly
```

```python
regression = LinearRegression()
regression.fit(x_train_poly, y_train)

plt.scatter(x_train, regression.predict(x_train_poly), color='red')
plt.scatter(x_train, y_train, color='blue')
plt.xlabel('xdataset')
plt.ylabel('ydataset')
plt.title('Polynomial Regression')
plt.show()
```

```python
x_train_poly
```

```python
from sklearn.metrics import r2_score
regression=LinearRegression()
regression.fit(x_train_poly, y_train)
y_pred=regression.predict(x_test_poly)
score=r2_score(y_test,y_pred)
print("R2 score for polynomial regression is:",score)
```

```python
## predictin for new dataset

x_new=np.linspace(-3,3,200).reshape(200,1)
x_new_poly=poly.transform(x_new)
x_new_poly

```

```python
y_new=regression.predict(x_new_poly)
plt.plot(x_new,y_new,color='red',label='prediction')
plt.plot(x_train,y_train,'o',color='blue',label='training data')
plt.plot(x_test,y_test,'o',color='green',label='testing data')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Polynomial Regression')
plt.legend()
plt.show()
```

         PIPELINE CONCEPT

```python
from sklearn import pipeline
```

```python
def poly_regression(degree):
    X_new = np.linspace(-3, 3, 200).reshape(200, 1)
    
    poly_features=PolynomialFeatures(degree=degree,include_bias=True)
    lin_reg=LinearRegression()
    poly_regression=Pipeline([
        ("poly_features",poly_features),
        ("lin_reg",lin_reg)
    ])
    poly_regression.fit(X_train,y_train) ## ploynomial and fit of linear reression
    y_pred_new=poly_regression.predict(X_new)
    #plotting prediction line
    plt.plot(X_new, y_pred_new,'r', label="Degree " + str(degree), linewidth=2)
    plt.plot(X_train, y_train, "b.", linewidth=3)
    plt.plot(X_test, y_test, "g.", linewidth=3)
    plt.legend(loc="upper left")
    plt.xlabel("X")
    plt.ylabel("y")
    plt.axis([-4,4, 0, 10])
    plt.show()
    
```
